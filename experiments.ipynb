```python
 ##combine one hot encoder collumns with the original data

import pandas as pd

data = pd.read_csv("Churn_Modelling.csv")

geo_encoded_df = pd.get_dummies(data['Geography'], prefix='Geography')

data = pd.concat(
    [data.drop('Geography', axis=1), geo_encoded_df],
    axis=1
)

data.head()
```

```python
##preprocess the data
## drop irrelevant columns
data = data.drop(['RowNumber', 'CustomerId', 'Surname'], axis=1, errors='ignore')
data
```

```python
from sklearn.preprocessing import LabelEncoder

label_encoder_gender = LabelEncoder()

data['Gender'] = label_encoder_gender.fit_transform(data['Gender'])

print(data.head())
```
